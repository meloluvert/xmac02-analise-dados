### Questão 1
Realize um teste de hipótese que verifique se desenvolvedores que utilizam Python possuem remuneração média anual diferente daqueles que não utilizam Python. Realize o teste com confiança de 95%.

Dica: use \
dados_q1 = df.dropna(subset=["salary_usd_yearly", "uses_python"]).copy() \
para gerar uma cópia limpa de NaN nas colunas de interesse.

h0: média _anual py = media_anual_sem_py 

h1: média anual py <> média anual sem py


````python


df = pd.read_csv('stackoverflow_2024.csv')
# print(df.head())
dados_q1 = df.dropna(subset=["salary_usd_yearly", "uses_python"]).copy()


programadores_com_py = dados_q1[dados_q1['uses_python'] == True]['salary_usd_yearly']
programadores_sem_py = dados_q1[dados_q1['uses_python'] == False]['salary_usd_yearly']
````

````python
weightstats.ttest_ind(programadores_com_py,programadores_sem_py, equal_var=False)
````

pvalue bem menor que 0.05, reiejiramos a hiótesel nula, as médias sãk diferentes

### Questão 2
Realize um teste de hipótese para investigar se a proporção de profissionais com remuneração acima da mediana é igual entre trabalhadores remotos e trabalhadores não remotos.

Utilize as variáveis:

```text
salary_above_median
is_remote

Lembre-se de limpar NaN nas colunas de interesse. Realize o teste com 95% de confiança.

````python
dados_q2 = df.dropna(subset=["salary_above_median", "is_remote"]).copy()


nao_remotos = dados_q2[dados_q2["is_remote"] == False]
remotos = dados_q2[dados_q2["is_remote"] == True]

print("Não remotos:", len(nao_remotos))
n_a = (nao_remotos["salary_above_median"] == True).sum()
print("Não remotos acima da mediana:",n_a)

print("Remotos:", len(remotos))
n_ab = (remotos["salary_above_median"] == True).sum()
print("Remotos acima da mediana:",n_ab )



proportion.test_proportions_2indep(n_a, len(nao_remotos),n_ab,len(remotos), method='score')
````
