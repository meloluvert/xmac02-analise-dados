## Estatística descritiva 
### Média 
Obter a média de um conjunto de valores requer operações aritméticas básicas. \
Exemplo: Obtenha a média dos valores 1, 2, 3, 3

```python
lista = [1,2,3,3]
media = sum(lista)/len(lista)
media
```

Existe uma função capaz de calcular diretamente a média de um conjunto?

```python
import statistics as st
import matplotlib.pyplot as plt
```

```python
st.mean(lista)
```

Agora podemos calcular a mediana e a moda.

```python
st.median(lista)
```

```python
st.mode(lista           )
```

Como calcular a amplitude?

```python
max(lista) - min(lista)
```

Como obter o desvio padrão?

```python

#standart desviation
st.stdev(lista)
```

Como obter a variância?

```python
st.variance(lista)
```

Como obter os quartis?

```python
q= st.quantiles(lista)
q
```

Como obter a amplitude interquartil?

```python
q[2] - q[0]
```

Vamos agora trabalhar com o dataset Tips (Gorjetas). Esse dataset contém dados das contas de clientes de um restaurante, informando o total da conta, o valor da gorjeta, além de outros dados.

```python
import seaborn as sns   
```

```python
df = sns.load_dataset('tips')
```

```python
df.head()
```

Qual o valor médio das gorjetas?

```python
#pega todas as gorjetas
df['tip'].mean()
```

Qual é a mediana das gorjetas?

```python
df['tip'].median()
```

Qual é o desvio padrão das gorjetas?

```python
df['tip'].std()
```

Vamos agora obter os quartis das gorjetas.

```python
q3 = st.quantiles(df['tip'])
q3
```

Vamos agora gerar um boxplot do valor das gorjetas.

```python
df.boxplot(column='tip')
plt.show()
```

É possível gerar boxplots dos dados de uma coluna agrupados pelos dados de outra. Por exemplo, podemos gerar um boxplot das gorjetas por dia da semana, ou por sexo do pagante.

```python
df.boxplot(by='day', column='tip')
plt.show()
```

```python
df.boxplot(by='sex', column='tip')
plt.show()
```

```python
#EXEMPLO DE BOX PLOT COM HUE
# 1. Carrega o dataset de exemplo embutido no Seaborn
df = sns.load_dataset("tips")

# 2. Configura a área de plotagem para ter 2 gráficos lado a lado (1 linha, 2 colunas)
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# 3. Gráfico 1: SEM hue
# Analisa a distribuição do valor da conta (total_bill) apenas por dia (day)
sns.boxplot(
    data=df, 
    x="day", 
    y="total_bill", 
    ax=axes[0], 
    palette="pastel"
)
axes[0].set_title("Sem 'hue': Conta total por Dia")
axes[0].set_ylabel("Conta Total (US$)")
axes[0].set_xlabel("Dia da Semana")

# 4. Gráfico 2: COM hue
# Analisa a conta por dia (day), mas subdivide cada dia entre Fumantes e Não Fumantes (smoker)
sns.boxplot(
    data=df, 
    x="day", 
    y="total_bill", 
    hue="smoker", # Adiciona a subdivisão
    ax=axes[1], 
    palette="pastel"
)
axes[1].set_title("Com 'hue': Conta por Dia, subdividida por Fumante")
axes[1].set_ylabel("Conta Total (US$)")
axes[1].set_xlabel("Dia da Semana")

# Ajusta o layout para não sobrepor textos e exibe o gráfico
plt.tight_layout()
plt.show()
```
