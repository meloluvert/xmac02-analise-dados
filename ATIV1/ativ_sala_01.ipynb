### Questão 4
O dataset "real_state_ny.csv" mostra dados de casas à venda no estado de Nova Iorque, nos EUA. Utilizando este dataset: \
a) Informe a média, a mediana e o desvio padrão do preço das casas de 3 quartos (bed = 3) da cidade de Albany (city = 'Albany') \
b) Gere um boxplot do preço do imóvel por número de quartos na cidade de Albany. Limite o número de quartos a no máximo 5.

```python
print(df[(df['bed'] == 3) & (df['city'] == 'Albany')]['price'].describe())

df[(df['bed'].isin(range(0,6))) & (df['city'] == 'Albany')].boxplot(by='bed', column='price')
```

### Questão 5
Plote um gráfico de pizza que mostre a porcentagem de imóveis por número de quartos para a cidade de Albany.

```python
df_beds_albany = df[(df['city'] == 'Albany')]

#for numero in df[]
contagem_por_num_quarto= df_beds_albany.groupby('bed')[['status']].count()

print(contagem_por_num_quarto.index)

plt.pie(     
    contagem_por_num_quarto['status'],
    labels= contagem_por_num_quarto.index,
    autopct='%1.1f%%'
)
```

### Questão 6
Plote um gráfico de barras que mostre o preço médio das casas de acordo com o número de banheiros nas cidades de New York, Albany, Yonkers e Long Beach. Limite o número de banheiros a 4, ou seja, descarte os imóveis que tenham mais que 4 banheiros.

```python
df_cities = df[(df['city'].isin(['New York','Albany', 'Yonkers' , 'Long Beach']))]
df_limit_bath = df_cities[df_cities['bath'].isin(range(0,5))]
sns.set_style('white')
sns.barplot(x='city', y='price', hue='bath', data=df_limit_bath)
```

### Questão 7
Plote dois histogramas lado a lado comparando o preço dos imóveis nas cidades de Yokers e Albany. Use bins=15.

```python
df_yonkers = df[df['city'] == 'Yonkers']
df_albany = df[df['city'] == 'Albany']

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(data=df_yonkers, x='price', bins=15, ax=axes[0])
axes[0].set_title('Yonkers')
axes[0].set_xlabel('Preço')

sns.histplot(data=df_albany, x='price', bins=15, ax=axes[1])
axes[1].set_title('Albany')
axes[1].set_xlabel('Preço')

plt.tight_layout()
```
