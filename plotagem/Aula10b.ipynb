## Criação de histogramas com Seaborn
Nesta aula analisaremos o índice de percepção de corrupção no setor público nos diferentes países do globo.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('datasets/CPI_2014.csv', encoding='latin-1')
```

```python
df
```

```python
sns.displot(df['CPI_2014'], kde=True)
sns.set_style('darkgrid')
plt.title('indice de percepção corrupção (2014)')

plt.xlabel("nível de corrupção")
plt.axvline(60, color='k',linestyle='--')
plt.show()
```

Exercício \
Criar um histograma apenas com países da Europa.

```python
df2 = df[df['Continent'] == 'Europe']
sns.displot(df2['CPI_2014'], kde=True)
sns.set_style('darkgrid')
plt.title('indice de percepção corrupção (2014)')

plt.xlabel("nível de corrupção")
plt.axvline(60, color='k',linestyle='--')
plt.show()
```

```python
sns.displot(df2['CPI_2014'], bins=15, kde=True)
sns.set_style('darkgrid')
plt.title('indice de percepção corrupção (2014)')

plt.xlabel("nível de corrupção")
plt.axvline(60, color='k',linestyle='--')
plt.show()
```
