## Gráficos de Dispersão
Vamos criar um gráfico de dispersão usando dados de idade e pressão arterial. A ideia é observar como a pressão varia em função da idade.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('datasets/bloodPressure.csv')
```

Vamos agora criar o gráfico, iniciando com a parametrização de dimensões, estilos, tamanhos de fonte, etc.

```python
plt.rcParams['figure.figsize']=[8,6]
sns.set_context("notebook", font_scale=1.2, rc={"font.size":12, 'axes.titlesize':16, 'axes.labelsize':16})
sns.set_style('darkgrid')
```

```python
#fit reg é a reta
sns.regplot(x='Age', y='SBP', fit_reg=False,data=df )
plt.title('Relationsip betewwem age and sbp')
plt.xlabel('Age of patiants')
plt.ylabel('SBP of patients')
```

Notamos a existência de um outlier com pressão 220. Existem algumas técnicas para lidar com outliers. Vamos criar um segundo dataframe extraindo o outlier.

```python
df_sem_outlier = df[df['SBP'] !=220]
```

```python
#fit reg é a reta
sns.regplot(x='Age', y='SBP', fit_reg=True,data=df_sem_outlier )
plt.title('Relationsip betewwem age and sbp')
plt.xlabel('Age of patiants')
plt.ylabel('SBP of patients')
```
