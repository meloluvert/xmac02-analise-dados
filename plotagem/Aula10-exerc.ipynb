1) Exiba histogramas lado a lado mostrando o índice de corrupção na Europa e na África. Use bins=15.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('datasets/CPI_2014.csv', encoding='latin-1')
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(
    df[df['Continent'] == 'Europe']['CPI_2014'],
    kde=True,
    bins=15,
    ax=axes[0]
)

sns.histplot(
    df[df['Continent'] == 'Africa']['CPI_2014'],
    kde=True,
    bins=15,
    ax=axes[1]
)

plt.show()
```
