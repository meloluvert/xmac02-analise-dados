Vamos agora utilizar o dataset LifeExpectancy para criar um gráfico de barras.

```python
df_life = pd.read_csv('datasets/lifeExpectancy.csv')
df_life.head()
```

Nosso gráfico deve permitir comparar a expectativa de vida de homens e mulheres de países africanos.

```python
df_life.plot(x='Countries', kind='barh', stacked=False)
plt.title("Expectativa de vida nos países africanos")
plt.xlabel('Países')
plt.ylabel('Expectativa de vida')

plt.show()
```

Para uma melhor visualização, podemos dividir o dataset. A seguir vamos tomar os 15 primeiros países do dataset.

```python
df_life_2 = df_life.head(15)
df_life_2.plot(x='Countries', kind='barh', stacked=False)
plt.title("Expectativa de vida nos países africanos")
plt.xlabel('Países')
plt.ylabel('Expectativa de vida')

plt.show()
```
